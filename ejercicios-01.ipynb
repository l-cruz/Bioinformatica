## Ejercicio 1. Replicación del ADN

**Secuencia inicial:**

- Hebra codificante: 5′-ATG CCG TTA GCT-3′
- Hebra complementaria: 3′-TAC GGC AAT CGA-5′

**Una ronda de replicación.** La replicación es semiconservativa: cada molécula hija contiene una hebra original y otra nueva.

| Molécula hija | Hebra 5′ → 3′ | Hebra 3′ → 5′ |
|---|---|---|
| 1 | Original: ATG CCG TTA GCT | Nueva: TAC GGC AAT CGA |
| 2 | Nueva: ATG CCG TTA GCT | Original: TAC GGC AAT CGA |

**Enzimas implicadas:**

- **Helicasa:** separa las dos hebras al romper los puentes de hidrógeno entre sus bases.
- **Primasa:** sintetiza cebadores cortos de ARN para iniciar la síntesis.
- **ADN polimerasa:** incorpora nucleótidos complementarios en dirección 5′ → 3′ y puede corregir ciertos errores.
- **Ligasa:** une los fragmentos de ADN, especialmente los de Okazaki en la hebra rezagada.

**Reflexión.** Si un error de replicación escapa a la reparación y queda fijado, se produce una mutación. Puede transmitirse a las células hijas y afectar a un producto génico, aunque también puede resultar silenciosa o afectar a una región no codificante.


### Extensión

```python
from Bio.Seq import Seq

adn_secuencia = Seq("ATGCCGTTAGCT")

hebra_complementaria = adn_secuencia.complement()
hebra_inversa_complementaria = adn_secuencia.reverse_complement()

print(f"Hebra original (5'->3'):       {adn_secuencia}")
print(f"Hebra complementaria (3'->5'): {hebra_complementaria}")
print(f"Hebra complementaria inversa (5'->3'): {hebra_inversa_complementaria}")

# Comprobación de que coincide con el cálculo manual: TACGGCAATCGA
```

## Ejercicio 2. Transcripción del ADN a ARN

**Secuencias de partida:**

- Hebra codificante: 5′-ATG CCT GAA TGC-3′
- Hebra molde: 3′-TAC GGA CTT ACG-5′

La ARN polimerasa lee la **hebra molde** de 3′ a 5′ y sintetiza ARN de 5′ a 3′. El ARNm resultante es **5′-AUG CCU GAA UGC-3′**: coincide con la hebra codificante, sustituyendo T por U.

**Promotor y región codificante.** El promotor ayuda a reclutar la maquinaria de transcripción y determina dónde se inicia. La región codificante es la parte del transcrito que se traduce a proteína. Se puede transcribir ARN fuera de la región codificante (por ejemplo, las UTR), así que ambos conceptos no son equivalentes.

**Cambio de orientación.** Si se toma la hebra opuesta como codificante, se obtiene otro ARNm, `GCAUUCAGGCAU`. Ese cálculo ilustra la orientación, pero no significa que el gen original se transcriba necesariamente desde ambos sentidos.


```python
from Bio import SeqIO
from pathlib import Path

fasta_data = ">secuencia_ejercicio_2\nATGCCTGAATGC\n"
ruta_archivo = Path("ejercicio2_real.fasta")
ruta_archivo.write_text(fasta_data, encoding="utf-8")

registro = SeqIO.read(ruta_archivo, "fasta")
adn_codificante = registro.seq

arnm = adn_codificante.transcribe()
adn_codificante_opuesto = adn_codificante.reverse_complement()
arnm_orientacion_opuesta = adn_codificante_opuesto.transcribe()

print(f"ADN codificante (5′→3′): {adn_codificante}")
print(f"ARNm original (5′→3′):  {arnm}")
print(f"ARNm si la hebra opuesta fuese codificante: {arnm_orientacion_opuesta}")
```

## Ejercicio 3. Traducción del ARNm a proteína

**ARNm:** 5′-AUG UAU GCU UAA-3′.

| Codón | Significado |
|---|---|
| AUG | Inicio; metionina (Met) |
| UAU | Tirosina (Tyr) |
| GCU | Alanina (Ala) |
| UAA | Parada; no codifica un aminoácido |

**Resultado:** Met–Tyr–Ala. En Biopython, `MYA*` incluye el símbolo `*` para señalar la parada.

**Si AUG mutase a GUG:** en el código genético estándar, GUG codifica valina. En una célula eucariota, ese cambio puede impedir el inicio de la traducción en esa posición; también podría iniciarse en otro AUG apropiado. No basta con sustituir Met por Val sin considerar cómo se inicia la traducción.

**Si desapareciese UAA:** el ribosoma podría continuar leyendo hasta otro codón de parada en el mismo marco, alterando la longitud y posiblemente la función de la proteína.


```python
from Bio.Seq import Seq

arnm_secuencia = Seq("AUGUAUGCUUAA")

proteina = arnm_secuencia.translate()

print(f"Secuencia ARNm: {arnm_secuencia}")
print(f"Secuencia Proteica: {proteina}")
```

## Ejercicio 4. Splicing alternativo

**Gen hipotético:** exones 1–2–3–4–5. Por ejemplo, dos transcritos alternativos podrían contener los exones **1–2–4–5** (isoforma A) o **1–3–5** (isoforma B), siempre que sus sitios de empalme sean compatibles.

Al incluir distintos exones, los transcritos pueden producir proteínas diferentes, aunque también pueden diferir en regiones no traducidas o dar lugar a ARN que no produce proteína. No podemos deducir dominios, localización o función concreta sin conocer las secuencias y los marcos de lectura.

**Ejemplo real: FGFR2.** En este gen, el uso alternativo de los exones IIIb y IIIc cambia parte del dominio extracelular y puede modificar la especificidad de unión a ligandos. Ensembl muestra distintos transcritos anotados; el número total de transcritos **no equivale** al número demostrado de proteínas funcionales ni permite atribuir todas las diferencias exclusivamente al splicing.

**Reflexión:** el splicing alternativo permite que un gen contribuya a diversos productos de ARN y, en algunos casos, proteínas distintas. El resultado depende del tejido y de la regulación celular.


```python
import requests

def consultar_transcritos_ensembl(gen="FGFR2"):
    """Consulta las anotaciones actuales; necesita conexión a Ensembl."""
    url = f"https://rest.ensembl.org/lookup/symbol/homo_sapiens/{gen}"
    try:
        respuesta = requests.get(
            url,
            params={"expand": 1},
            headers={"Accept": "application/json"},
            timeout=15,
        )
        respuesta.raise_for_status()
        datos = respuesta.json()
    except (requests.RequestException, ValueError) as error:
        print(f"No se pudo consultar Ensembl: {error}")
        return

    transcritos = datos.get("Transcript", [])
    print(f"Gen: {datos.get('display_name')} | ID: {datos.get('id')}")
    print(f"Transcritos anotados actualmente: {len(transcritos)}")
    print("Este total no representa necesariamente proteínas distintas ni solo eventos de splicing.")
    for transcrito in transcritos[:5]:
        print(transcrito.get("id"), transcrito.get("display_name"), transcrito.get("length"))

consultar_transcritos_ensembl()
```

## Ejercicio 5. Introducción a las proteínas

**Secuencia:** Met–Ile–Ser–Gly–Val–Lys–His. El extremo **N-terminal** es Met y el **C-terminal** es His.

El orden de los aminoácidos determina qué interacciones son posibles y condiciona el plegamiento y la función de la proteína. El entorno celular también influye: no es correcto afirmar que la secuencia garantiza por sí sola una única conformación en toda circunstancia.

**Cambio de un residuo hidrofóbico a uno polar:** si el residuo queda enterrado en el núcleo, la sustitución podría alterar su estabilidad hidrofóbica, forzando a la proteína a desestabilizarse o plegarse de forma anómala para exponer la nueva carga al medio acuoso. 

**Extensión (GFP, PDB `1EMA`):** Esta proteína tiene una característica estructura de barril beta (hojas beta organizadas en un cilindro) que envuelve y protege el cromóforo fluorescente en su interior (donde se aprecian hélices alfa). A continuación, descargamos el archivo PDB y lo renderizamos interactivamente para observar visualmente esta conformación.


```python
from Bio.PDB import PDBList
import py3Dmol

try:
    archivo_pdb = PDBList(verbose=False).retrieve_pdb_file(
        "1EMA", pdir=".", file_format="pdb"
    )
    print(f"Estructura 1EMA descargada exitosamente en: {archivo_pdb}")
    
    with open(archivo_pdb, "r") as f:
        pdb_data = f.read()
        
    view = py3Dmol.view(width=800, height=400)
    view.addModel(pdb_data, "pdb")
    
    view.setStyle({'cartoon': {'color': 'spectrum'}})
    view.zoomTo()
    
    print("\nVisualización 3D: Arrastra para rotar. Observa la gran estructura cilíndrica de láminas beta (barril) rodeando el centro.")
    view.show()

except Exception as error:
    print(f"No se pudo descargar o renderizar la estructura: {error}")
```

## Ejercicio 6. Actividad integradora: del ADN a la proteína

Para realizar el recorrido completo del dogma central de la biología molecular, en lugar de usar una secuencia aleatoria, utilizaremos un caso real: el **gen de la insulina humana (registro NM_000207.3)**. Descargaremos su secuencia directamente desde la base de datos Nucleotide del NCBI utilizando el módulo `Entrez` de Biopython.

1. **Replicación:** Aislamos la secuencia codificante (CDS) extraída del registro. En el pipeline, generamos las nuevas moléculas hijas comprobando cómo cada una conserva una hebra original (molde o codificante) y sintetiza la nueva hebra complementaria.
2. **Transcripción:** A partir de la nueva hebra molde, la ARN polimerasa sintetiza el ARNm (sustituyendo Timina por Uracilo).
3. **Traducción:** El ARNm se traduce en la cadena de aminoácidos correspondiente a la insulina, deteniendo la lectura automáticamente al encontrar un codón de parada.

**Reflexión sobre vulnerabilidad a errores:** 
El punto más crítico y vulnerable es la **replicación**. Si un error (mutación) en el ADN escapa a los mecanismos de reparación, esta alteración se vuelve permanente y se transmitirá al genoma de las células hijas. Por consiguiente, todas las moléculas de ARNm y proteínas que se generen a partir de ese gen defectuoso estarán alteradas. Por el contrario, un error puntual durante la transcripción o traducción generará un lote transitorio de ARN o proteínas defectuosas que terminarán degradándose, pero el molde original (el ADN) seguirá intacto para futuras lecturas.

```python
from Bio import Entrez, SeqIO

Entrez.email = "lucia.cruz104@alu.ulpgc.es" 

print("Descargando el registro GenBank de la Insulina Humana (NM_000207.3)...")
with Entrez.efetch(db="nucleotide", id="NM_000207.3", rettype="gb", retmode="text") as handle:
    registro = SeqIO.read(handle, "genbank")

cds_secuencia = None
for feature in registro.features:
    if feature.type == "CDS":
        cds_secuencia = feature.extract(registro.seq)
        break
        
# Procesamos la secuencia CDS al completo sin recortarla
secuencia = cds_secuencia

def pipeline_dogma_central(adn):
    # Acortamos el print a 60 caracteres para que el notebook quede limpio, pero procesamos toda la cadena
    print(f"\n[ORIGINAL] ADN Codificante (5'->3') [Total: {len(adn)} pb]:\n{adn[:60]}... (truncado visualmente)")
    
    adn_molde = adn.complement()
    print(f"\n[PASO 1: REPLICACIÓN]")
    print(f"Molécula hija 1 (Conserva codificante original):")
    print(f" 5'->3' : {adn[:60]}...")
    print(f" 3'->5' : {adn_molde[:60]}")
    print(f"Molécula hija 2 (Conserva molde original):")
    print(f" 5'->3' : {adn[:60]}")
    print(f" 3'->5' : {adn_molde[:60]}...")
    
    arnm = adn.transcribe()
    print(f"\n[PASO 2: TRANSCRIPCIÓN] ARNm (5'->3'):\n{arnm[:60]}...")

    proteina = arnm.translate(to_stop=True)
    print(f"\n[PASO 3: TRADUCCIÓN] Proteína (Preproinsulina completa) [{len(proteina)} aa]:\n{proteina}")
    
    return proteina

resultado = pipeline_dogma_central(secuencia)
```
